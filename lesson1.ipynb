NumPy Arrays 
</h1>

```python
import numpy as np
arr = np.array([1,2,3,4,5,6], dtype=float)
print(arr)
```

```python
arr = np.array([[1,2,3],[1,2,3]])
print(arr)
```

```python
arr = np.zeros([5,5])
print(arr)
```

```python
arr = np.ones([5,5])
print(arr)
```

```python
arr = np.empty(1)
print(arr)
```

```python
arr = np.full([2,3],7)
print(arr)
```

```python
arr = np.eye(7)
print(arr)
```

```python
arr = np.identity(7)
print(arr)
```

```python
arr = np.arange(0,10,2)
print(arr)
```

```python
arr = np.linspace(0,10,5)
print(arr)
```

```python
arr = np.logspace(0,10,3)
print(arr)
```
